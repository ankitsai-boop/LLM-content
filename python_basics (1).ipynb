### <font color="purple">**Section 1: Variables, Data Types & Type Conversion**</font>
- Variables store data values.
- Key basic data types: `int` (integers), `float` (decimals), `str` (text), and `bool` (True/False).

```python
# 1. Variable Assignment
user_name = "Alex"
age = 25
hourly_rate = 45.50
is_enrolled = True

# 2. Checking Data Types
print("Type of user_name:", type(user_name))
print("Type of age:", type(age))
print("Type of hourly_rate:", type(hourly_rate))
print("Type of is_enrolled:", type(is_enrolled))

# 3. Type Conversion (Casting)
age_str = str(age)          # Convert int to string
rate_int = int(hourly_rate)  # Convert float to integer (truncates decimal)

print(f"\nFormatted Output: {user_name} is {age_str} years old and earns ${hourly_rate}/hr.")
print(f"Truncated Integer Rate: ${rate_int}")
```

### <font color="purple">**Section 2: String Formatting & Methods**</font>
- Use `f-strings` (`f"{variable}"`) for dynamic string interpolation.
- Common string operations: `.strip()`, `.upper()`, `.lower()`, `.replace()`, and `.split()`.

```python
raw_title = "  data analyst career track  "

# Cleaning string data
clean_title = raw_title.strip().title()
uppercase_title = clean_title.upper()
words_list = clean_title.split()

print("Cleaned Title:", clean_title)
print("Uppercase:", uppercase_title)
print("Split into Words:", words_list)
print("Replaced Word:", clean_title.replace("Data", "Business"))
```

### <font color="purple">**Section 3: Core Data Structures (Lists, Tuples, Dictionaries, Sets)**</font>
- **List (`[]`)**: Ordered, mutable sequence.
- **Tuple (`()`)**: Ordered, immutable sequence.
- **Dictionary (`{key: value}`)**: Key-value lookup mapping.
- **Set (`{}`)**: Unordered collection of unique items.

```python
# Lists (Mutable )
skills = ["Python", "SQL", "Excel"]
skills.append("Tableau")
print("Updated Skills List:", skills)

# Dictionaries (Key-Value Pairs)
student_profile = {
    "name": "Jordan",
    "courses": ["Python 101", "Math 201"],
    "gpa": 3.8
}
student_profile["gpa"] = 3.9  # Update key
student_profile["graduated"] = False  # Add new key
print("Student Profile:", student_profile)

# Sets (De-duplication)
duplicate_ids = [101, 102, 101, 105, 103, 102]
unique_ids = set(duplicate_ids)
print("Unique IDs:", unique_ids)
```

### <font color="purple">**Section 4: Control Flow & Loops**</font>
- Decision making using `if`, `elif`, and `else`.
- Iteration using `for` loops, `while` loops, and concise **List Comprehensions**.

```python
scores = [78, 92, 45, 88, 62, 95]

# 1. Conditional Logic & For Loop
print("--- Evaluation Results ---")
for score in scores:
    if score >= 90:
        grade = 'A'
    elif score >= 75:
        grade = 'B'
    elif score >= 60:
        grade = 'C'
    else:
        grade = 'F'
    print(f"Score: {score} -> Grade: {grade}")

# 2. List Comprehension (Extract only scores >= 75)
passing_scores = [s for s in scores if s >= 75]
print("\nPassing Scores List:", passing_scores)
```

### <font color="purple">**Section 5: Reusable Functions**</font>
- Modularize code block execution with parameter passing and return statements.

```python
def calculate_total(price: float, tax_rate: float = 0.08, discount: float = 0.0) -> float:
    """Calculates total price after tax and optional discount."""
    discounted_price = price * (1 - discount)
    total_price = discounted_price * (1 + tax_rate)
    return round(total_price, 2)

# Calling function with default tax_rate
item1 = calculate_total(100.0)
# Calling function with specified parameters
item2 = calculate_total(200.0, tax_rate=0.10, discount=0.15)

print(f"Item 1 Final Total: ${item1}")
print(f"Item 2 Final Total: ${item2}")
```

### <font color="purple">**Section 6: Basic Exception Handling**</font>
- Prevent script crashes when encountering bad input or runtime errors using `try-except`.

```python
def safe_divide(a, b):
    try:
        result = a / b
        return f"Result: {result}"
    except ZeroDivisionError:
        return "Error: Cannot divide by zero!"
    except TypeError:
        return "Error: Both inputs must be numerical values!"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "two"))
```

### <font color="purple">**Section 7: Practical Mini-Project (Employee Pay Summary)**</font>
**Objective:** Process a list of employee work logs, calculate net wages, apply bonuses for overtime (>40 hrs), and summarize key statistics.

```python
# Dataset: List of Employee Records
employees = [
    {"name": "Alice", "hours": 45, "rate": 30.0},
    {"name": "Bob", "hours": 38, "rate": 25.0},
    {"name": "Charlie", "hours": 50, "rate": 40.0},
    {"name": "Diana", "hours": 40, "rate": 28.0}
]

def process_payroll(data):
    payroll_summary = []
    total_payout = 0
    
    for emp in data:
        name = emp['name']
        hours = emp['hours']
        rate = emp['rate']
        
        # Calculate regular and overtime pay (1.5x for hours > 40)
        if hours > 40:
            reg_pay = 40 * rate
            overtime_pay = (hours - 40) * (rate * 1.5)
            total_pay = reg_pay + overtime_pay
        else:
            total_pay = hours * rate
            
        payroll_summary.append({"name": name, "total_pay": total_pay})
        total_payout += total_pay
        
    return payroll_summary, total_payout

# Run Payroll Processor
summary, grand_total = process_payroll(employees)

print("=== PAYROLL REPORT ===")
for item in summary:
    print(f"Employee: {item['name']:<10} | Pay: ${item['total_pay']:.2f}")
print("-" * 36)
print(f"Grand Total Expenditure: ${grand_total:.2f}")
```
